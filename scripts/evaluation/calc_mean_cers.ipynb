```python
def open_and_concat_seen_dfs(model, languages):
    model_list = model.split("_")
    target = model_list[0]
    strat = model_list[1]

    dfs = []
    for l in languages:
        if strat != "rand":
            file_path = f"ASR_ICLDC/Results/{model}_{l}_seen_results.tsv"
        else:
            file_path = f"ASR_ICLDC/Results/rand_{l}_seen_results.tsv"
        
        df = pd.read_csv(file_path, sep="\t")
        df["Language"] = l
        df["Model"] = model
        df["Target_Language"] = target
        df["Strategy"] = strat

        dfs.append(df)
    concat_df = pd.concat(dfs, ignore_index=True)
    return concat_df
```

```python
shf_df = open_and_concat_seen_dfs("hsb_fam", ["cs", "pl", "sk", "bg"])
shg_df = open_and_concat_seen_dfs("hsb_geo", ["cs", "pl", "sk", "sl"])
shp_df = open_and_concat_seen_dfs("hsb_phon", ["bg", "lt", "ro", "uk"])
slf_df = open_and_concat_seen_dfs("lg_fam", ["rw", "sw", "bas", "yo"])
slg_df = open_and_concat_seen_dfs("lg_geo", ["luo", "rw", "sw", "bas"])
slp_df = open_and_concat_seen_dfs("lg_phon", ["luo", "nl", "id", "bas"])
stf_df = open_and_concat_seen_dfs("tt_fam", ["ba", "ky", "ug", "uz"])
stg_df = open_and_concat_seen_dfs("tt_geo", ["cv", "ru", "ba", "et"])
stp_df = open_and_concat_seen_dfs("tt_phon", ["ba", "hi", "ug", "pl"])
sar_df = open_and_concat_seen_dfs("all_rand", ["sv-SE", "uk", "ab", "ro"])
```
